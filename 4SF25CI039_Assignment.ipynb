Question 1: Create a Pandas DataFrame with 3 columns and 5 rows filled with random integers. Add a new column that is the product of the first two columns.

```python
data={
    'a':[4,38,52,67,71],
    'b':[3,7,8,9,10],
    'c':[9,11,22,29,54]
}
df=pd.DataFrame(data)
print(df)
df['Product']=df['a']*df['b']
print(df)
```

Question 3: Create a Pandas DataFrame with 3 columns and 5 rows filled with random integers. Introduce some NaN values. Fill the NaN values with the mean of the respective columns.

```python
data={
    'a':[2,64,np.nan,71,82],
    'b':[4,5,7,10,np.nan],
    'c':[np.nan,17,22,36,52]
}
df=pd.DataFrame(data)
print(df)
df['a']=df['a'].fillna(df['a'].mean())
df['b']=df['b'].fillna(df['b'].mean())
df['c']=df['c'].fillna(df['c'].mean())
print(df)
```

Question 5: Create a Pandas DataFrame with 2 columns: 'Category' and 'Value'. Fill the 'Category' column with random categories ('A', 'B', 'C') and the 'Value' column with random integers. Group the DataFrame by 'Category' and compute the sum and mean of 'Value' for each category.

```python
data={
    'Category':['A','B','C','A','C','B'],
    'Values':[5,6,8,25,41,56]
}
df=pd.DataFrame(data)
print(df)
grouped=df.groupby('Category')['Values']
print(grouped.sum())
print(grouped.mean())
```

Question 6: Create a Pandas DataFrame with 3 columns: 'Product', 'Category', and 'Sales'. Fill the DataFrame with random data. Group the DataFrame by 'Category' and compute the total sales for each category.

```python
data={
    'Products': ['Harry Potter', 'Notebook', 'Mobile', 'Headphones','Black Forest', 'Atomic Habits', 'Cheesecake'],
    'Category': ['Books', 'Stationery', 'Electronics', 'Electronics','Cake', 'Books', 'Cake'],
    'Sales': [450, 200, 45000, 2500, 850, 600, 950]
}
df=pd.DataFrame(data)
grouped=df.groupby('Category')['Sales'].sum()
print(grouped)
```

Question 7: Create two Pandas DataFrames with a common column. Merge the DataFrames using the common column.

```python
Student_info={
    'Id':[2,4,6,8],
    'Names':['Hari','Sushma','Aarya','Kushal']
}
student_df=pd.DataFrame(Student_info)
print(student_df)
Marks={
    'Id':[2,4,6,8],
    'marks':[77,79,82,94]
}
mark_df=pd.DataFrame(Marks)
print(mark_df)
merge_df=pd.merge(student_df,mark_df,on='Id')
print(merge_df)
```

Question 8: Create a Pandas DataFrame with a MultiIndex (hierarchical index). Perform some basic indexing and slicing operations on the MultiIndex DataFrame.

Question 9: Create a Pandas DataFrame with MultiIndex consisting of 'Category' and 'SubCategory'. Fill the DataFrame with random data and compute the sum of values for each 'Category' and 'SubCategory'.

Question 10: Create a Pandas DataFrame with 3 columns and 5 rows filled with random integers. Apply a function that doubles the values of the DataFrame.

```python
data={
    'A':[4,11,32,17,43,10],
    'B':[13,60,30,68,41,47],
    'C':[17,19,32,24,46,45]
}
df=pd.DataFrame(data)
print(df)
def double(n):
    return n*2
df=df.apply(double)
print(df)
```

Question 11: Create a Pandas DataFrame with 3 columns and 6 rows filled with random integers. Apply a lambda function to create a new column that is the sum of the existing columns.

```python
data={
    'A':[3,7,9,15,11,30],
    'B':[99,36,12,77,45,20],
    'C':[17,22,16,26,40,41]
}
df=pd.DataFrame(data)
print(df)
df['Sum']=df.apply(lambda x: x['A']+x['B']+x['C'],axis=1)
print(df)
```
